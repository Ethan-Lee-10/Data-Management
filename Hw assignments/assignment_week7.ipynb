### Exercise 1 (1 point):

Create a histogram using the "age" column.

```python
df["age"].plot(kind="hist", 
               bins=32,
               title="Age of Survey-Taker",
               xlabel="Age Ranges",
               ylabel="Frequency"
              )
```

### Exercise 2 (1 point):

Create a bar plot showing the counts of the "regions" in the data.

```python
df_value_counts = df["region"].value_counts()
```

```python
df_value_counts.plot(
    kind="bar",
    title="Region of Survey-Taker",
    xlabel="Region",
    ylabel="Frequency"
)
```

### Exercise 4 (1 point):

Create a pie plot showing the number of smokers vs non-smokers in the data.

```python
_ = df["smoker"].value_counts().plot(kind="pie",
                                    title="Are They a Smoker?",
                                    )
```

### Exercise 5 (1 point):

Create a horizontal stacked bar plot on the combined group of "region" and "smoker".

```python
cross_tab_df = pd.crosstab(df["region"],
                           df["smoker"],
                          )
```

```python
_ = cross_tab_df.plot(kind="barh", stacked=True, figsize=(8,3), title="Smokers in a region")
```
