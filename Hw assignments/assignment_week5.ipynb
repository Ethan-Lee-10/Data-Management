## CSVs Assignment

#### Instructions:

Download the provided products.csv file into the same directory as this notebook.

Read the products.csv file into a DataFrame.

Add a new column called "Total Value", calculated as Price * Stock.

Filter out products with a Total Value below 10000.

Save the resulting DataFrame to a new file called filtered_products.csv.  The resulting csv should contain exactly 5 columns.

````python
products_df = pd.read_csv("products.csv")
````

````python
products_df["Total Value"] = products_df["Price"] * products_df["Stock"]
````

````python
products_df = products_df[products_df["Total Value"] >= 10000]
````

````python
products_df
````

````python
products_df.to_csv("filtered_products.csv", index=False)
````

## JSON Assignment

#### Instructions:

Download the provided employees.csv file.

Read the employees.csv file into a DataFrame.

Convert the DataFrame into a JSON file called employees.json.

Read the employees.json file back into a new DataFrame.

Filter out employees with a salary above 50,000.

Save the filtered DataFrame to a new JSON file called high_salary_employees.json.

The output JSON should be of the following example structure (the whitespace does not matter):
```
{
    "2":{"Name":"Bob","Department":"Finance","Salary":52000},
    "4":{"Name":"David","Department":"Marketing","Salary":60000},
 ... etc.
 }
```
*Hints:*
  * What does this suggest about the index of the dataframe?
  * What are the relevant kwargs to the `.to_json()` method?

````python
def display_json(json_obj):
    """Pretty print ``json_obj``."""
    print(json.dumps(json.loads(json_obj), indent=2))
````

````python
employees_df = pd.read_csv("employees.csv")
````

````python
employees_df.to_json("employees.json", orient="index")
````

````python
employees_json = pd.read_json("employees.json", orient="index")
````

````python
filtered_df = employees_json[employees_json["Salary"] > 50000]
````

````python
filtered_df.to_json("high_salary_employees.json", orient="index")
````

````python
print(filtered_df)
````

````python
display_json(filtered_df.to_json(orient="index"))
````

## Excel Assignment

#### Instructions:
Download the provided sales.xlsx file.

Calculate the total sales for each product and add a new column called Total Sales.

Read both the CSV and JSON files back into new DataFrames to confirm the data integrity.

````python
sales_df = pd.read_excel("sales.xlsx")
````

````python
sales_df["Total Sales"] = sales_df["Units Sold"] * sales_df["Unit Price"]
````

````python
sales_df.to_csv("sales_summary.csv", index=False)
````

````python
sales_df.to_json("sales_summary.json", orient="records")
````
