```python
import pandas as pd
```

```python
mlb_df = pd.read_csv("./baseball.csv")
mlb_df
```

## Step 1 (1 point): Create three small datasets and concatenate them into one.

1. Create a DataFrame of all rows where the Year is 2012
2. Create a DataFrame of all rows where the Year is 2011
3. Create a DataFrame of all rows where the Year is 2010
4. concatenate those into one DataFrame.
5. Use ignore_index = True to ensure that the index is consistent 

```python
year_2012_mask = mlb_df["Year"] == 2012
year_2011_mask = mlb_df["Year"] == 2011
year_2010_mask = mlb_df["Year"] == 2010
```

```python
year_2012_df = mlb_df[year_2012_mask]
year_2011_df = mlb_df[year_2011_mask]
year_2010_df = mlb_df[year_2010_mask]
```

```python
mlb_df_slice = pd.concat(
    [year_2012_df, year_2011_df, year_2010_df], ignore_index=True
)
```

```python
mlb_df_slice
```

```python
team_names_df = pd.read_csv("./team_names.csv")
team_names_df
```

```python
df_with_full_name_slice = mlb_df_slice.merge(
    team_names_df,
    left_on="Team",
    right_on="Team ID",
    how="left"
)
```

```python
df_with_full_name_slice["Full Team Name"].isna().sum()
```

## Step 3 (1 point): Create 5 bins based on the the Batting Average (BA) column.  The bins should be roughly evenly spaced in terms of _frequency_.  Hint: Which binning function does that point to?

1. Save the value in a new column called "BA_bins"
2. Get the value counts for each of the bins.  Sort by the bins from lowest threshold to highest threshold.

```python
df_with_full_name_slice["BA_bins"] = pd.qcut(
    df_with_full_name_slice["BA"],
    q=5
)
```

```python
df_with_full_name_slice["BA_bins"].value_counts().sort_index()
```

## Step 4 (1 point):  Rank the teams in the highest of the 5 bins. The highest batting average is ranked #1.  Then use the code that follows to print out the team identifiers, and batting average, and rank in order.  

The code that follows assumes that you have a dataframe called `df_with_full_name_slice` with a column called `Full Name`, and that the dataframe is already sorted appropriately.  You may choose to modify the code, but it should not be necessary.

```python
highest_ba_bin = df_with_full_name_slice["BA_bins"].cat.categories[-1]

df_with_full_name_slice = df_with_full_name_slice[
    df_with_full_name_slice["BA_bins"] == highest_ba_bin
]
```

```python
df_with_full_name_slice["BA_rank"] = df_with_full_name_slice["BA"].rank(
    ascending=False
)
```

```python
df_with_full_name_slice = df_with_full_name_slice.sort_values("BA_rank")
```

```python
df_with_full_name_slice = df_with_full_name_slice.rename(
    columns={"Full Team Name": "Full Name"}
)
```

```python
for _, row in df_with_full_name_slice.iterrows():
    print(f"{row['Team']}-{row['League']} is the {row['Full Name']} "
          f"with a BA of {row['BA']} in {row['Year']} and rank {row['BA_rank']}")
```
