```python
df.isna().head()
```

```python
df.isna().sum()
```

```python
df = df.dropna(subset=["Email"])
```

```python
df = df.fillna({
    "Age": df["Age"].median(),
    "Salary": df["Salary"].median()
})
```

```python
print(df)
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

```python
# We use drop_duplicates() without any column header because if we were to pinpoint "Name" or "Email" duplicates, it could leave us with Names and emails that don't align, so we will just drop any complete duplicates to help narrow down the things to clean.
```

```python
df
```

```python
df["Name"] = df["Name"].str.strip().str.title()
```

```python
df["Email"] = df["Email"].str.strip().str.lower()
```

```python
df
```

```python
df["Email_Name"] = (
    df["Email"]
    .str.split("@").str[0]
    .str.title()
)
```

```python
df = df[df["Name"] == df["Email_Name"]]
```

```python
df = df.drop(columns=["Email_Name"])
```

```python
df
```

```python
df.to_csv("cleaned_data.csv", index=False)
```
